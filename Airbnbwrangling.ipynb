```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\monik\OneDrive\Desktop\DescStat\Data_Analysis_Workshops\data\airbnb.csv")

#dropping the accidental index column if it exists
if 'Unnamed: 0' in df.columns:
    df.drop('Unnamed: 0', axis=1, inplace=True)

df['coordinates'] = df['coordinates'].str.replace('(', '', regex=False).str.replace(')', '', regex=False)
coords = df['coordinates'].str.split(',', expand=True)
df['latitude'] = coords[0].astype(float)
df['longitude'] = coords[1].astype(float)
df.drop('coordinates', axis=1, inplace=True) # drop the old column


df['price'] = df['price'].str.replace('$', '', regex=False).astype(float)
df['price'] = df['price'].fillna(df['price'].mean())

df['listing_added'] = pd.to_datetime(df['listing_added'], errors='coerce')
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')


mappings = {
    'private room': 'Private Room',
    'private': 'Private Room',
    'entire home/apt': 'Entire place',
    'shared room': 'Shared room',
    'home': 'Entire place'
}

df['room_type'] = df['room_type'].str.lower().str.strip()
df['room_type'] = df['room_type'].replace(mappings)

neigh = df['neighbourhood_full'].str.split(',', n=1, expand=True)
df['borough'] = neigh[0].str.strip()
df['neighbourhood'] = neigh[1].str.strip()
df.drop('neighbourhood_full', axis=1, inplace=True) # drop the old column

df.loc[df['rating'] > 5.0, 'rating'] = 5.0

df.fillna({
    'reviews_per_month': 0,
    'number_of_stays': 0,
    '5_stars': 0
}, inplace=True)

df['is_rated'] = np.where(df['rating'].isna(), 0, 1)

inconsistent_dates = df[df['listing_added'] > df['last_review']]
df.drop(inconsistent_dates.index, inplace=True)

#drop exactly identical rows
df.drop_duplicates(inplace=True)

#saving the cleaned dataset to a new csv file without the index
df.to_csv('airbnb_cleaned.csv', index=False)

print( df.head())
```
